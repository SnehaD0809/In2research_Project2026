```python
import pandas as pd
import plotly.express as px
import plotly.io as pio

# Load your modelling dataset
df = pd.read_csv("ERIC_modeling_ready_imputed2.csv")

# Check the columns
print(df.columns)
```

```python
df[["latitude", "longitude"]].head()
```

```python
df[["latitude", "longitude"]].isna().sum()
```

```python
df_map = df.dropna(subset=["latitude", "longitude"])
```

Mapping for Age Vulnerability

```python
fig = px.density_map(
    df_map,
    lat="latitude",
    lon="longitude",
    z="age_vulnerability_score",
    animation_frame="year",
    map_style="open-street-map",
    zoom=5,
    radius=20,
    width=900,
    height=700,
    color_continuous_scale="Viridis"
)

fig.show()
```

Mapping for Flooding

```python
fig = px.density_map(
    df_map,
    lat="latitude",
    lon="longitude",
    z="flood_occurrences_triggering_a_risk_assessment_no",
    animation_frame="year",
    map_style="open-street-map",
    zoom=5,
    radius=20,
    width=900,
    height=700,
    color_continuous_scale="viridis"
)

fig.show()
```

Mapping for Overheating Occurrences

```python
fig = px.density_map(
    df_map,
    lat="latitude",
    lon="longitude",
    z="overheating_occurrences_triggering_a_risk_assessment_no",
    animation_frame="year",
    map_style="open-street-map",
    zoom=5,
    radius=20,
    width=900,
    height=700,
    color_continuous_scale="solar"
)

fig.show()
```

Spatial Mapping for Total Backlog costs

```python
fig = px.density_map(
    df_map,
    lat="latitude",
    lon="longitude",
    z="total_backlog_cost_gbp",
    animation_frame="year",
    map_style="open-street-map",
    zoom=5,
    radius=20,
    width=900,
    height=700,
    color_continuous_scale="Viridis"
)

fig.show()
```

```python
pio.write_html(
    fig,
    file="age_vulnerability_map.html",
    auto_play=True
)
```
